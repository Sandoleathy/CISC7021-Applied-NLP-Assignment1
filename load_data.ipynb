```python
from datasets import Dataset
import os

path = "ModelAndDatasets/data/TinyStories-train.txt"
stories, buf = [], []
with open(path, encoding="utf-8") as f:
    for line in f:
        line = line.strip()
        if line == "<|endoftext|>":
            if buf:
                stories.append(" ".join(buf)); buf = []
        elif line:
            buf.append(line)
if buf:
    stories.append(" ".join(buf))

print("total stories:", len(stories))
ds = Dataset.from_dict({"text": stories})
ds.save_to_disk("TinyStories-train-parsed")   # 只跑一次，以后直接 load_from_disk
```
